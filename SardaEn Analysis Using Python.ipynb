```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

#Load Data (e.g., wfrom a CSV file)
#Replace 'data.csv' with your actual file path
try:
    df = pd.read_csv('SardaEnergy.csv')
    print("Data loaded successfully!")
except FileNotFoundError:
    print("File not found. Ensure the path is correct.")
```

```python
#Data Inspection
print(df.head()) #View the first 5 rows
print(df.info()) #Check data types and missing values
print(df.describe()) #Get statistical summary (mean, median, etc.
```

```python
#Data Cleaning
#Example: Fill missing numeric balues with the column mean
df = df.fillna(df.mean(numeric_only=True))
```

```python
#Data Grouping & Analysis
#Example: Group by a category and calculate the mean of a value
grouped_data = df.groupby('Year')['Sales'].mean()
print(grouped_data)
```

```python
#Data Visualisation
sns.set_theme(style='whitegrid')
plt.figure(figsize=(10,6))
sns.barplot(x=grouped_data.index, y=grouped_data.values)
plt.title('Average Sales by Year')
plt.xlabel('Average Sales by Year')
plt.ylabel('Year')
plt.show()
```
